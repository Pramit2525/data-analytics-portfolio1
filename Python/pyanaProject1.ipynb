```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

df = pd.read_csv(
    r"C:\Users\prami\OneDrive\Desktop\Portfolio of DA\Portfolio Project 1\dat_academy_retail_sales.csv",
    parse_dates=["order_date"]
)

df.head()
```

📌 3. Data Quality Check & Preparation
Check structure & missing values


```python
df.info()
df.isnull().sum()
```

```python
Create calculated fields
```

```python
df["revenue"] = df["units_sold"] * df["unit_price"] * (1 - df["discount_pct"])
df["profit"] = df["revenue"] - df["cost"]
```

📌 4. KPI Metrics

```python
total_revenue = df["revenue"].sum()
total_profit = df["profit"].sum()
total_orders = df["order_id"].nunique()

print("Total Revenue:", total_revenue)
print("Total Profit:", total_profit)
print("Total Orders:", total_orders)
```

📌 5. Revenue Analysis

5.1 Revenue by Region

```python
rev_region = df.groupby("region")["revenue"].sum().sort_values(ascending=False)
rev_region
```

Monthly revenue trend

```python
# Convert order_date to datetime
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")

# Create month column
df["month"] = df["order_date"].dt.to_period("M")

# Monthly revenue
rev_month = df.groupby("month")["revenue"].sum()

plt.figure(figsize=(12,5))
rev_month.plot(kind="line", marker="o")
plt.title("Monthly Revenue Trend")
plt.ylabel("Revenue")
plt.xlabel("Month")
plt.show()
```

📌 6. Product Performance

6.1 Revenue by Category

```python
rev_category = df.groupby("product_category")["revenue"].sum().sort_values(ascending=False)
rev_category
```

6.2 Profit by Category

```python
profit_category = df.groupby("product_category")["profit"].sum().sort_values(ascending=False)
profit_category
```

6.3 Top 10 Products by Revenue

```python
top_products = df.groupby("product_category")["revenue"].sum().sort_values(ascending=False).head(10)
top_products
```

📌 7. Customer Analysis

7.1 Revenue by Customer Segment

```python
rev_segment = df.groupby("customer_segment")["revenue"].sum().sort_values(ascending=False)
rev_segment
```

7.2 Top 10 Customers by Revenue

```python
top_customers = df.groupby("customer_id")["revenue"].sum().sort_values(ascending=False).head(10)
top_customers
```

📌 8. Sales Channel Analysis

```python
rev_channel = df.groupby("sales_channel")["revenue"].sum().sort_values(ascending=False)
rev_channel
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=rev_channel.index, y=rev_channel.values)
plt.title("Revenue by Sales Channel")
plt.ylabel("Revenue")
plt.show()
```
